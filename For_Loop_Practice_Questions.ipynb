**Basic - For Loop**

1.	Print numbers from 1 to 20.
2.	Print all odd numbers between 1 and 50.
3.	Print the multiplication table of 9.
4.	Find the sum of numbers from 1 to 50.
5.	Print all elements of a list.
6.	Print a triangle pattern of stars:
7.	Calculate factorial of a number.
8.	Print all prime numbers between 1 and 50.
9.	Iterate through a string and print each character.
10.	Print numbers divisible by 3 between 1 and 100.



```python
## Print numbers from 1 to 20.

for i in range (1,21):
  print(i)
```

```python
## Print all odd numbers between 1 and 50.

for i in range (1,51,2):
  print(i)
```

```python
## Print the multiplication table of 9.

for i in range (9,91,9):
  print(i)
```

```python
## Find the sum of numbers from 1 to 50.

total = 0
for i in range(1,51):
  total = total + i

print("Sum of Numbers from 1 to 50: ", total)
```

```python
## Print all elements of a list.

my_list = ["Apple", "Orange", "Grapes", "Mango", "Banana"]
for i in my_list:
  print(i)
```

```python
## Print a triangle pattern of stars:

for i in range(1, 6):
    print("*" * i)
```

```python
## Calculate factorial of a number.

number = int(input("Enter the number: "))

factorial = 1

for i in range(1, number+1):
  factorial = factorial*i

print("Factorial of", number, "is", factorial)
```

```python
## Print all prime numbers between 1 and 50.

for num in range(2,51):
  is_prime = True

  for i in range(2,num):

    if num % i == 0:
      is_prime = False
      break

  if is_prime:
    print(num)
```

```python
## Iterate through a string and print each character.

word = "Python"

for character in word:
  print(character)
```

```python
## Print numbers divisible by 3 between 1 and 100.

for i in range(3,101,3):
  print(i)
```

**For Loop with break and continue**

1.	Print numbers 1 to 10 but stop when number divisible by 7 (break).
2.	Print numbers 1 to 20 but skip all odd numbers (continue).
3.	Iterate through a list of fruits and stop when "Mango" is found (break).
4.	Iterate through a string and skip vowels (continue).
5.	Print multiplication table of 5 but skip 5×3 (continue) and stop if product > 35 (break).
6.	Print numbers 1 to 50 but stop on first number divisible by 13 (break).
7.	Iterate a list of marks and skip marks < 35 (continue).
8.	Print squares of numbers 1 to 15 but stop if square > 100 (break).
9.	Iterate through numbers 1 to 30 and skip multiples of 4 (continue).
10.	Print factorial steps but stop if factorial > 1000 (break).

```python
## Print numbers 1 to 10 but stop when number divisible by 7 (break).

for i in range(1,11):
  if i % 7 == 0:
    break
  print(i)
```

```python
## Print numbers 1 to 20 but skip all odd numbers (continue).

for i in range(1,21):
  if i % 2 != 0:
    continue
  print(i)
```

```python
## Iterate through a list of fruits and stop when "Mango" is found (break).

my_list = ["Apple", "Orange", "Grapes", "Mango", "Banana"]

for fruit in my_list:
  if fruit == "Mango":
    break
  print(fruit)
```

```python
## Iterate through a string and skip vowels (continue).

word = "Reevaluate"
for consonants in word:
  if consonants.lower() in "aeiou":
    continue
  print(consonants)
```

```python
## Print multiplication table of 5 but skip 5×3 (continue) and stop if product > 35 (break).

for i in range(1,11):
  product = 5 * i

  if i == 3:
    continue #(skips the iteration)

  if product > 35:
    break

  print("5 *", i, "=", product)
```

```python
## Print numbers 1 to 50 but stop on first number divisible by 13 (break).

for i in range(1,51):
  if i % 13 == 0:
    break
  print(i)
```

```python
## Iterate a list of marks and skip marks < 35 (continue).

student_marks = [25,40,70,80,65,59,35,19]

for marks in student_marks:
  if marks < 35:
    continue
  print(marks)
```

```python
## Print squares of numbers 1 to 15 but stop if square > 100 (break).

for i in range(1,16):
  if i**2 > 100:
    break
  print(i**2)
```

```python
## Iterate through numbers 1 to 30 and skip multiples of 4 (continue).

for i in range(1,31):
  if i % 4 == 0:
    continue
  print(i)
```

```python
## Print factorial steps but stop if factorial > 1000 (break).

factorial = 1

for i in range(1, 11):
  factorial = factorial*i

  if factorial > 1000:
    break

  print(i, "factorial =", factorial)
```
